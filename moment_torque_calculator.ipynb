```python
# Moment and Torque Calculator
# M = r x F  (cross product)

import numpy as np
import matplotlib.pyplot as plt

# 2D moment calculation
# position vector r from point to force application (meters)
r = np.array([3.0, 2.0, 0.0])

# force vector F (Newtons)
F = np.array([4.0, -3.0, 0.0])

# moment M = r x F
M = np.cross(r, F)

# magnitude
mag_M = np.linalg.norm(M)

# direction
if M[2] > 0:
    direction = "Counter-Clockwise (CCW)"
else:
    direction = "Clockwise (CW)"

print("=== 2D Moment Calculator ===")
print("Position vector r: " + str(r) + " m")
print("Force vector F:    " + str(F) + " N")
print("Moment vector M:   " + str(M) + " N.m")
print("Magnitude:         " + str(round(mag_M, 3)) + " N.m")
print("Direction:         " + direction)
```

```python
# 3D Moment Calculator 
# position vector r (meters)
r_3d = np.array([2.0, 3.0, 1.0]) 

# force vector F (Newtons) 
F_3d =np.array([4.0, -2.0, 3.0]) 

# moment M = r x F 
M_3d = np.cross(r_3d, F_3d)

# magnitude 
mag_3d = np.linalg.norm(M_3d)

# unit vector of moment 
unit_M = M_3d / mag_3d 

print("=== 3D Moment Calculator ===") 
print("Position vector r: " +str(r_3d) + " m") 
print("Force vector F: " + str(F_3d) + " N") 
print("Moment vector M: " + str(M_3d) + " N.m")
print("Magnitude: " + str(round(mag_3d, 3)) + " N.m") 
print("Unit vector: " + str(np.round(unit_M, 3)))
```

```python
# Moment Diagram

fig, ax = plt.subplots(figsize=(8, 8))

# plot position vector r
ax.annotate('', xy=r[:2], xytext=[0,0],
            arrowprops=dict(arrowstyle='->', color='steelblue', lw=2))
ax.text(r[0]*1.05, r[1]*1.05, 'r = ' + str(r[:2]) + ' m',
        color='steelblue', fontsize=10)

# plot force vector F from tip of r
ax.annotate('', xy=[r[0]+F[0], r[1]+F[1]], xytext=r[:2],
            arrowprops=dict(arrowstyle='->', color='darkorange', lw=2))
ax.text(r[0]+F[0]*0.5, r[1]+F[1]*0.5, 'F = ' + str(F[:2]) + ' N',
        color='darkorange', fontsize=10)

# plot moment arc
theta = np.linspace(0, np.pi/2, 100)
arc_r = 0.8
arc_x = arc_r * np.cos(theta)
arc_y = arc_r * np.sin(theta)
if M[2] > 0:
    ax.plot(arc_x, arc_y, color='green', lw=2)
    ax.annotate('M = ' + str(round(mag_M, 2)) + ' N.m CCW',
                xy=(0.2, 0.9), fontsize=10, color='green')
else:
    ax.plot(arc_x, -arc_y, color='red', lw=2)
    ax.annotate('M = ' + str(round(mag_M, 2)) + ' N.m CW',
                xy=(0.2, -0.9), fontsize=10, color='red')

# origin point
ax.plot(0, 0, 'ko', markersize=8)
ax.text(0.1, -0.3, 'O (moment point)', fontsize=9)

ax.set_xlim(-2, 8)
ax.set_ylim(-4, 6)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.grid(True, alpha=0.3)
ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
ax.set_title('Moment and Force Diagram')
plt.tight_layout()
plt.savefig('moment_diagram.png')
plt.show()
print("diagram saved")
```
